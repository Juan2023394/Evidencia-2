```python
df = pd.read_csv('Datos_actividad_2.csv')

df.info()
```

```python
df['Ganancia'] = 0
df['Ganancia'] = df['Ingresos'] - df['Gastos']
df['Margen porcentual']=0
df['Margen porcentual'] = df['Ganancia'] / df['Ingresos'] * 100
```

```python
sucursal = df.loc[df['Gastos'] == df['Gastos'].max()]
sucursal
```

```python
df['Fecha'] = pd.to_datetime(df['Fecha'])
df['Mes'] = df['Fecha'].dt.month
resumen = df.groupby('Mes').agg({'Ingresos':'sum', 'Gastos':'sum'})
resumen
```

```python
top5 = resumen.nlargest(5, 'Ingresos')
resumen['Top 5'] = resumen.index.isin(top5.index)
resumen
```

```python
resumen['Estrategia'] = ''
resumen.loc[resumen['Gastos'] > resumen['Ingresos'].mean(), 'Estrategia'] = 'Reducir gastos'
resumen.to_excel('estrategias.xlsx')
resumen
```
